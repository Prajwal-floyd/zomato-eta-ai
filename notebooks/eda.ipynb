```python
import pandas as pd

df = pd.read_csv("../data/train.csv")

ordered_clean = (
    df['Time_Orderd']
    .apply(normalize_time)
)

picked_clean = (
    df['Time_Order_picked']
    .apply(normalize_time)
)

print(ordered_clean.head(20).tolist())

print(picked_clean.head(20).tolist())
```

```python
ordered_time = pd.to_datetime(
    ordered_clean,
    format='%H:%M',
    errors='coerce'
)

picked_time = pd.to_datetime(
    picked_clean,
    format='%H:%M',
    errors='coerce'
)

prep_time = (
    picked_time - ordered_time
).dt.total_seconds() / 60

prep_time = prep_time.apply(
    lambda x: x + 1440
    if pd.notnull(x) and x < 0
    else x
)

print(prep_time.head(20))
```
